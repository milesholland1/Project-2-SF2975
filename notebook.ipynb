```
from matplotlib import pyplot as plt
import numpy as np
import math

m = 2
# Generate m random data points along 1 dimensions
Z = np.array(np.random.normal(0, 1, m))
print(Z)
```

```
# Now define the stock price
s0 = 100
r = 0.03
sigma = 0.2
T = 1

S_T = s0*np.exp((r-(sigma**2)/2)*T + sigma*math.sqrt(T)*Z)

print(S_T)
```

```
# Here we want to solve the Monte Carlo estimator
K = 100

V = math.exp(-r*T)*(1/m)*(sum(S_T-K))

print(V)
```
